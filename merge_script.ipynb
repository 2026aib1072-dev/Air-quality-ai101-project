```python
import os
import nbformat

# 1. Define team files in exact execution sequence
notebook_sequence = [
    {"file": "01_data_preprocessing_vineet_updated.ipynb", "author": "Vineet", "section": "1: Data Preprocessing"},
    {"file": "02_eda_and_modeling_rishabh.ipynb", "author": "Rishabh", "section": "2: Exploratory Data Analysis & Baseline Modeling"},
    {"file": "04_Model_evaluation_and_visualizations.ipynb", "author": "Divya Godara", "section": "3: Advanced Model Evaluation & Visualizations"},
    {"file": "5_odds_ratios_and_borderline_analysis_sandeep.ipynb", "author": "Sandeep", "section": "4: Odds Ratios & Borderline Analysis"}
]

output_master_file = "air_quality_master_pipeline.ipynb"

# 2. Create new master notebook structure
master_nb = nbformat.v4.new_notebook()

# 3. Add Master Title Cell
title_markdown = """# Air Quality Prediction & Hazardous Day Classification
## Master Integrated Pipeline

**Authors:** Vineet, Rishabh, Divya Godara, Sandeep
**Dataset:** `city_day.csv`
***
"""
master_nb.cells.append(nbformat.v4.new_markdown_cell(title_markdown))

# 4. Merge notebook cells sequentially
merged_count = 0
for item in notebook_sequence:
    fname = item["file"]

    if os.path.exists(fname):
        print(f"Processing: {fname}...")

        # Add section header divider cell
        section_divider = f"""---
## Section {item['section']}
**Author:** {item['author']}
**Source Notebook:** `{fname}`
---"""
        master_nb.cells.append(nbformat.v4.new_markdown_cell(section_divider))

        # Read notebook cells
        with open(fname, 'r', encoding='utf-8') as f:
            nb = nbformat.read(f, as_version=4)
            master_nb.cells.extend(nb.cells)

        merged_count += 1
    else:
        print(f"⚠️ Warning: File '{fname}' not found!")

# 5. Save unified master pipeline notebook
with open(output_master_file, 'w', encoding='utf-8') as f:
    nbformat.write(master_nb, f)

print(f"\n✅ Successfully merged {merged_count}/{len(notebook_sequence)} notebooks into '{output_master_file}'!")
```
